**Projeto 1: A Agenda de Contatos:**<br>
• **Contexto:** Você precisa criar um sistema simples para armazenar e gerenciar contatos (nome,
telefone e e-mail).<br>
• **Problema:** Como você pode adicionar novos contatos, buscar um contato pelo nome e listar
todos os contatos?<br>
• **Indicação de como resolver:** Pensar em uma estrutura que permita associar informações
(telefone e e-mail) a um nome de forma eficiente.<br>
• **Resolução do Projeto 1:** Um dicionário em Python é uma excelente escolha para este projeto.
A chave pode ser o nome do contato, e o valor pode ser outro dicionário contendo o telefone e
o e-mail.

```python
def adicionarContato(contatos):
  nome = input("\nDigite o nome do contato: ")
  telefone = input("Digite o telefone do contato: ")
  email = input("Digite o email do contato: ")
  contatos[nome] = [telefone, email]

def buscarContato(contatos):
  nome = input("\nDigite o nome do contato que deseja buscar: ")
  print(contatos.get(nome))
```

```python
contatos = {}
while True:
  print(f"""\n
  1 - Adicionar novo contato
  2 - Buscar contato
  3 - Listar todos contatos
  4 - Sair""")
  while True:
    try:
      opcao = int(input('Escolha uma opção: '))
      break
    except ValueError:
      print("Isso não é um número válido!")

  match opcao:
    case 1:
      adicionarContato(contatos)
    case 2:
      buscarContato(contatos)
    case 3:
      print(contatos)
    case 4:
      print("Encerrado")
      break
    case _:
        print("Escolha uma opção valida")
```

**Projeto 2: A Lista de Tarefas:** <br>
• **Contexto:** Você precisa criar uma lista de tarefas para a semana.<br>
• **Problema:** Como você pode adicionar novas tarefas, marcar tarefas como concluídas e visualizar as tarefas pendentes?<br>
• **Indicação de como resolver:** Pensar em uma forma de manter uma sequência de itens que podem ser adicionados e removidos.<br>
• **Resolução do Projeto 2:** Uma lista em Python é ideal para essa situação. Podemos adicionar novas tarefas ao final da lista, remover tarefas quando concluídas e simplesmente imprimir a lista para visualizar as tarefas pendentes.

```python
def adicionarTarefa(tarefas):
  adicionar = input("Digite o nome da tarefa: ")
  tarefas.append(adicionar)
  print("Tarefa Adicionada")

def marcarConcluido(tarefas):
  tarefa = input("Digite o nome da tarefa concluida: ")
  if tarefa in tarefas:
    tarefas.remove(tarefa)
    print("\nTarefa removida")
  else:
    print("\nNão existe tarefa com esse nome na lista")

def visualizarTarefa(tarefas):
  print(f"Tarefas Pendentes\n{tarefas}")
```

```python
tarefas = []
while True:
  print(f"""\n
  1 - Adicionar nova tarefa
  2 - Marcar tarefa com concluida
  3 - Visualizar tarefas pendentes
  4 - Sair""")
  while True:
    try:
      opcao = int(input('Escolha uma opção: '))
      break
    except ValueError:
      print("Isso não é um número válido!")

  match opcao:
    case 1:
      adicionarTarefa(tarefas)
    case 2:
      marcarConcluido(tarefas)
    case 3:
      visualizarTarefa(tarefas)
    case 4:
      print("Encerrado")
      break
    case _:
        print("Escolha uma opção valida")
```
