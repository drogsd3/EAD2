3 - **(Prática)** Implementar DFS em Python: Dado o grafo abaixo, representado por lista de adjacências, escreva uma função dfs(grafo, inicio) que retorne os vértices em ordem de descoberta pela busca em profundidade.<br>
Execute a partir de 'A'.

```python
#Grafo de exemplo
G = {
 'A': ['B', 'C'],
 'B': ['A', 'D', 'E'],
 'C': ['A', 'F'],
 'D': ['B'],
 'E': ['B', 'F'],
 'F': ['C', 'E']
}
```

```python
def dfs(grafo, inicio, visitados=None):
    if visitados is None:
        visitados = []
    visitados.append(inicio)
    for vizinho in grafo[inicio]:
        if vizinho not in visitados:
            visitados = dfs(grafo, vizinho, visitados)
    return visitados
```

```python
#Teste Saida Esperada ["A","B","D","E","F","C"]
ordem = dfs(G, "A")
print(ordem)
```

4 - **(Prática)** Implementar BFS em Python: Usando o mesmo grafo do Exercício 3, escreva uma função bfs(grafo, inicio) que retorne a ordem de descoberta
pela busca em largura e execute-a a partir de 'A'.

```python
from collections import deque
def bfs(grafo, inicio):
  visitados = [inicio]
  fila = deque([inicio])
  ordem = []
  while fila:
    vertice = fila.popleft()
    ordem.append(vertice)
    for vizinho in grafo[vertice]:
      if vizinho not in visitados:
        visitados.append(vizinho)
        fila.append(vizinho)
  return ordem
```

```python
#Teste Saida Esperada ['A', 'B', 'C', 'D', 'E', 'F']
ordem = bfs(G, "A")
print(ordem)
```
