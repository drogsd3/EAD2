5. **(Prática)** Implemente em Python uma função decimal_para_hexa(n) que converta um número decimal não negativo n para sua representação hexadecimal como string, usando uma pilha. (Por exemplo,
decimal_para_hexa(26) deve retornar "1A".)<br>

````python
from os.path import join
def decimal_para_hexa(n):
  pilha = []
  if n < 0:
    return None
  if n == 0:
    return "0"
  hexadecimal = "0123456789ABCDEF"
  while n > 0:
    pilha.append(hexadecimal[n % 16])
    n //= 16
  return "".join(pilha)
````

````python
print(decimal_para_hexa(26))
print(decimal_para_hexa(255))
print(decimal_para_hexa(0))
print(decimal_para_hexa(4095))
````

6. **(Prática)** Escreva uma função balanceia_colchetes(expressao) que, dada uma string contendo (, ), {, } e outros caracteres, determine se os parênteses e chaves estão balanceados. Use pilhas no seu algoritmo. Por exemplo,
"({[]})" está balanceado, mas "({[}])" não está.


````python
class Stack:
  def __init__(self):
    self.stack = []

  def push(self, element):
    self.stack.append(element)

  def pop(self):
    if self.is_empty():
      return "Stack is empty"
    return self.stack.pop()

  def peek(self):
    if self.is_empty():
      return "Stack is empty"
    return self.stack[-1]

  def is_empty(self):
    return len(self.stack) == 0

  def size(self):
    return len(self.stack)
````

````python
def balanceia_colchetes(expressao):
  pilha = Stack()
  for caractere in expressao:
    if caractere in '([{':
      pilha.push(caractere)
    elif caractere in ')]}':
      if pilha.is_empty():
        return False
      abertura = pilha.pop()
      if abertura == '(' and caractere != ')':
        return False
      if abertura == '[' and caractere != ']':
        return False
      if abertura == '{' and caractere != '}':
        return False
  return pilha.is_empty()
````

````python
print(balanceia_colchetes("({[]})"))
print(balanceia_colchetes("({[}])"))
print(balanceia_colchetes("{(a+b)*[c-d]}"))
print(balanceia_colchetes("((a+b)"))
````

**Projeto 1 – Histórico de Filtros em Dashboard de Dados**<br><br>
**Contexto:** Em aplicações de ciência de dados para negócios, usuários interagem com dashboards aplicando filtros e transformações aos dados (por exemplo, em tabelas ou gráficos interativos). É útil permitir que o usuário
**desfaça** filtros aplicados (por exemplo, revertendo ao estado anterior dos dados) sem recarregar tudo do início.<br><br>
**Problema:** Como implementar um mecanismo de **histórico de operações** que guarde estados anteriores dos dados ao aplicar filtros, para que o usuário possa desfazer a última ação?


````python
import pandas as pd
class HistoricoOperacoes:
  def __init__(self, dataframe):
    self.estado_atual = dataframe.copy()
    self.pilha_estados = Stack()

  def aplicar_filtro(self, condicao):
    self.pilha_estados.push(self.estado_atual.copy())
    self.estado_atual = self.estado_atual.loc[condicao].copy()

  def desfazer(self):
    if self.pilha_estados.is_empty():
      print("Nenhum estado anterior para desfazer.")
      return False
    self.estado_atual = self.pilha_estados.pop()
    return True
  def mostrar(self):
    print(self.estado_atual.to_string(index=False))
````

````python
# Exemplo
df = pd.DataFrame({
 "produto": ["Notebook", "Mouse", "Monitor", "Teclado"],
 "vendas": [3200, 800, 2100, 1200]
})

historico = HistoricoOperacoes(df)
historico.aplicar_filtro(historico.estado_atual["vendas"] >= 1000)
print("Após o filtro:")
historico.mostrar()
historico.desfazer()
print("\nApós desfazer:")
historico.mostrar()
````

**Projeto 2 – Calculadora de Expressões Aritméticas**<br><br>
**Contexto:** Muitas calculadoras e programas de matemática interpretam expressões aritméticas infixas (como "3 + 4 * (2 - 1)"). Uma maneira clássica de avaliar tais expressões é converter para notação pós-fixa (RPN) ou usar
pilhas para operar em tempo de execução.<br><br>
**Problema:** Escrever um programa Python que avalie uma expressão matemática em infixo contendo +, -, *, /, e
parênteses. Queremos usar pilhas para lidar com a precedência de operadores e a aninhamento de parênteses.

````python
import re

def avaliar_expressao(expressao):
  precedencia = {'+': 1, '-': 1, '*': 2, '/': 2}
  valores = Stack()
  operadores = Stack()

  tokens = re.findall(r'\d+(?:\.\d+)?|[()+\-*/]', expressao.replace(' ', ''))

  def aplicar_operador():
    operador = operadores.pop()
    b = valores.pop()
    a = valores.pop()

    if a is None or b is None:
      raise ValueError("Expressão inválida")

    if operador == '+':
      valores.push(a + b)
    elif operador == '-':
      valores.push(a - b)
    elif operador == '*':
      valores.push(a * b)
    elif operador == '/':
      if b == 0:
        raise ZeroDivisionError("Divisão por zero")
      valores.push(a / b)

  for token in tokens:
    if token.replace('.', '', 1).isdigit():
      valores.push(float(token) if '.' in token else int(token))

    elif token == '(':
      operadores.push(token)

    elif token == ')':
      while not operadores.is_empty() and operadores.peek() != '(':
        aplicar_operador()

      if operadores.is_empty():
        raise ValueError("Parênteses desbalanceados")
      operadores.pop()

    else:
        while (not operadores.is_empty() and operadores.peek() != '(' and precedencia[operadores.peek()] >= precedencia[token]):
           aplicar_operador()
        operadores.push(token)

  while not operadores.is_empty():
    if operadores.peek() == '(':
      raise ValueError("Parênteses desbalanceados")
    aplicar_operador()

  resultado = valores.pop()
  if not valores.is_empty():
    raise ValueError("Expressão inválida")
  return resultado
````

````python
print(avaliar_expressao("3 + 4 * (2 - 1)"))
print(avaliar_expressao("10 / 2 + 6 * 3"))
````
