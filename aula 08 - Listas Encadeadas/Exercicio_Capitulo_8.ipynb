**Exercício Prático 1:** Tamanho de uma lista<br>
Escreva uma função para encontrar o comprimento de uma lista
encadeada simples.<br>
Chame-a de: get_length() e como parâmetro deve ser passada a lista
encadeada.

````python
def get_length(lista_encadeada):
    current = lista_encadeada.head
    tamanho = 0
    while current is not None:
        tamanho += 1
        current = current.next
    return tamanho
````

**Exercício Prático 2:** Histórico de Navegação<br>
Implemente uma lista encadeada que simula o histórico de navegação de um navegador.<br> A lista deve permitir:<br>
1. Adicionar uma nova página ao histórico.
2. Exibir o histórico completo.
3. Voltar uma página no histórico (remover o último elemento)

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class NavegacaoHistorico:
    def __init__(self):
        self.head = None

    def adicionar_pagina(self, url):
        """Adiciona uma nova página ao final do histórico."""
        novo_no = Node(url)
        if not self.head:
            self.head = novo_no
            return

        atual = self.head
        while atual.next:
            atual = atual.next
        atual.next = novo_no

    def exibir_historico(self):
        """Exibe o histórico completo."""
        if not self.head:
            print("Histórico vazio.")
            return []

        historico = []
        atual = self.head
        while atual:
            historico.append(atual.data)
            atual = atual.next
        print(" -> ".join(historico))
        return historico

    def voltar_pagina(self):
        """Volta uma página no histórico (remove o último elemento)."""
        if not self.head:
            print("Nenhuma página para voltar.")
            return None

        # Se houver apenas um elemento
        if not self.head.next:
            removido = self.head.data
            self.head = None
            return removido

        atual = self.head
        while atual.next.next:
            atual = atual.next

        removido = atual.next.data
        atual.next = None
        return removido

# Exemplo de uso:
historico = NavegacaoHistorico()
historico.adicionar_pagina("google.com")
historico.adicionar_pagina("github.com")
historico.adicionar_pagina("stackoverflow.com")

print("Histórico Inicial:")
historico.exibir_historico()

print("\nVoltando uma página:")
historico.voltar_pagina()
historico.exibir_historico()
````
