R: Bubble Sort, Insertion sort e Merge Sort, são as ordenações estáveis estudadas, essa estabilidade se dá quando o algoritmo analisa a posição no vetor, verificará se é necessário fazer a substituição, evitando reposicionamento desnecessários.

7 - Considere X = [7, 4, 5, 2]. Trace o passo a passo de como o Bubble Sort ordenaria X em ordem crescente, mostrando o estado do vetor após cada passagem completa.<br>
Vetor inicial: [7, 4, 5, 2]<br>
**Passagem 1**<br>
• 7 e 4: troca → [4, 7, 5, 2]<br>
• 7 e 5: troca → [4, 5, 7, 2]<br>
• 7 e 2: troca → [4, 5, 2, 7]<br>
Após a 1ª passagem: [4, 5, 2, 7].<br>
O 7 chegou à posição final.<br>
**Passagem 2**<br>
• 4 e 5: não troca → [4, 5, 2, 7]<br>
• 5 e 2: troca → [4, 2, 5, 7]<br>
Após a 2ª passagem: [4, 2, 5, 7].<br>
**Passagem 3**<br>
• 4 e 2: troca → [2, 4, 5, 7]<br>
• 4 e 5: não troca.<br>
Após a 3ª passagem: [2, 4, 5, 7]. <br>
O vetor já está ordenado.
