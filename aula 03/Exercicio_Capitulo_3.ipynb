4. (PRÁTICA): Implemente uma função em Python chamada buscar_cliente(nome, lista_clientes) que realiza busca sequencial em uma lista de nomes de clientes. Essa função deve retornar o índice onde o nome foi
encontrado ou -1 caso o nome não esteja na lista_clientes. Em seguida, escreva um pequeno código de exemplo que utilize essa função para buscar por pelo menos um nome que exista e um que não exista na lista,
exibindo mensagens adequadas.

```python
def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1
```

```python
lista_clientes = ["João", "Maria", "Pedro", "Ana", "Carlos"]
nome = input("Digite o nome do cliente: ")
indice = buscar_cliente(nome, lista_clientes)
print(f'O indice é {indice}')
```

```python
nome = input("Digite o nome do cliente: ")
indice = buscar_cliente(nome, lista_clientes)
if indice == -1:
    print(f'O cliente {nome} não está na lista')
else:
    print(f'O cliente {nome} está na lista no indice {indice}')
```

5. (PRÁTICA): Crie uma função conta_ocorrencias(valor, lista) em Python que percorre uma lista e conta quantas vezes um determinado valor aparece. Utilize busca sequencial (varrendo a lista) para realizar a contagem. Por exemplo, conta_ocorrencias(3, [1,3,5,3,7]) deve retornar 2, pois o valor 3 aparece duas
vezes na lista. Inclua casos de teste chamando a função com diferentes valores e listas

```python
def conta_ocorrencias(valor, lista):
    contador = 0
    for i in lista:
        if i == valor:
            contador += 1
    return contador
```

```python
lista = [1,3,5,7,9,12,34,65,78,98,97,1,5,7,54,98]
valor = int(input("Digite um numero para procurar na lista: "))
print(f'O numero {valor} aparece {conta_ocorrencias(valor, lista)} vezes na lista')
```

6. (PRÁTICA): Suponha que precisamos remover um elemento específico de uma lista não ordenada. Escreva uma função Python remove_elemento(lista, valor) que remove a primeira ocorrência de valor na lista, mantendo os demais elementos e retornando o valor removido. A função deve retornar None caso o valor não
seja encontrado ou a lista esteja vazia. Dica: Utilize a busca sequencial para encontrar a posição do elemento e, se encontrada, deslize os elementos posteriores para “preencher” o espaço removido (ou use métodos próprios da lista Python). Demonstre o funcionamento da função com um exemplo.


```python
def remove_elemento(lista, valor):
    if valor in lista:
      lista.remove(valor)
      return valor
    else:
      return None
```

```python
lista = [1,3,5,7,9,12,34,65,78,98,97,1,5,7,54,98]
valor = int(input("Digite um numero para procurar na lista: "))
if remove_elemento(lista, valor) == None:
  print(f'O numero {valor} não foi encontrado na lista')
else:
  print(f'O numero {valor} foi removido da lista')
```

**Projeto 1:** Encontrando seu Filme Favorito
<br>
**Contexto:** Imagine um sistema de recomendação de filmes simples onde os filmes são armazenados em uma lista não-ordenada.
<br>
**Problema:** Implementar uma função que permita ao usuário buscar um filme pelo título nessa lista.
<br>
**Indicação de Solução:** Crie uma lista de títulos de filmes e uma função que use a busca sequencial para encontrar um título inserido pelo usuário. Se o filme for encontrado, exiba uma mensagem informando o título e sua posição (índice) na lista. Se não for encontrado, exiba uma mensagem informando que o filme não está no
catálogo.


```python
filme = ["Duro de matar", "Odisseia", "Filme do Pelé", "Velozes e Furiosos"]

def buscar_filme(filme, titulo):
    if titulo in filme:
        return print(f"O filme foi encontrado e está no indice {filme.index(titulo)}")
    else:
        return print(f'O filme {titulo} não está na lista')

#solução usando a indicação de solução
def buscar_filme_sequencial(filme, titulo):
    for i in range(len(filme)):
        if filme[i].lower() == titulo.lower():
            return print(f"O filme foi encontrado e está no indice {i}")
        else:
            return print(f'O filme {titulo} não está na lista')
```

```python
titulo = input("Digite o titulo do filme: ")
print(buscar_filme(filme, titulo))
```

```python
titulo = input("Digite o titulo do filme: ")
print(buscar_filme_sequencial(filme, titulo))
```

Projeto 2: Verificando a Lista de Convidados<br>.

**Contexto:** Desenvolver um programa para gerenciar a lista de convidados de um evento.
<br>
**Problema:** Implementar uma funcionalidade que permita verificar se um nome específico está na lista de convidados.
<br>
**Indicação de Solução:** Crie uma lista de nomes de convidados. Implemente uma função que receba um nome como entrada e use a busca sequencial para verificar se esse nome existe na lista. Exiba uma mensagem indicando se o convidado está na lista ou não.

```python
lista_convidados = ["João", "Maria", "Pedro", "Ana", "Carlos"]

def verificar_convidado(lista_convidados, nome):
    if nome in lista_convidados:
        return print(f"O/A {nome} está na lista")
    else:
        return print(f"O/A {nome} não está na lista")

def verificar_convidado_sequencial(lista_convidados, nome):
    for i in range(len(lista_convidados)):
        if lista_convidados[i] == nome:
            return print(f"O {nome} está na lista")
        else:
            return print(f"O {nome} não está na lista")
```

```python
nome = input("Digite o nome do convidado: ")
verificar_convidado(lista_convidados, nome)
```

```python
nome = input("Digite o nome do convidado: ")
verificar_convidado_sequencial(lista_convidados, nome)
```
