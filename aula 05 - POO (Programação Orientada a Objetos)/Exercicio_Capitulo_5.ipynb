**5. (PRÁTICA): Classe Pessoa:**
<br>Crie uma classe Pessoa com nome e idade. Adicione método
faz_aniversario() que incrementa a idade em 1. Use __str__() para retornar algo como "nome, idade, anos". Crie duas pessoas e mostre a idade de cada uma antes e depois de chamar faz_aniversario().

````python
class Pessoa:
  def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade

  def __str__(self):
    return f"Nome: {self.nome}, {self.idade} anos"

  def faz_aniversario(self):
    self.idade += 1
````

````python
pessoa1 = Pessoa("Ana", 28)
pessoa2 = Pessoa('João', 19)
print(pessoa1)
print(pessoa2)
````

````python
pessoa1.faz_aniversario()
pessoa2.faz_aniversario()
print(pessoa1)
print(pessoa2)
````

**Projeto 1: Biblioteca de Livros**
<br>
**Contexto do mundo real:** Imagine um sistema para gerenciar o empréstimo de livros em uma pequena biblioteca escolar. Queremos cadastrar livros e usuários, registrar quando alguém empresta ou devolve um livro.
<br>
**Descrição do problema:** Cada Livro tem título, autor, quantidade disponível. Cada Usuário tem nome, matrícula e lista de livros emprestados. O sistema deve permitir que um usuário faça empréstimos (reduzindo a quantidade disponível) e devoluções (aumentando quantidade). Devemos evitar empréstimos se o livro
estiver indisponível. Também precisamos manter os dados organizados e seguros (ex.: não permitir mexer diretamente na quantidade disponível).

````python
class Livro:
  def __init__(self, titulo, autor, quantidade):
    self.__titulo = titulo
    self.__autor = autor
    self.__quantidade = quantidade

  def __str__(self):
      return f"{self.__titulo} escrito por {self.__autor}, {self.__quantidade} unidade(s) disponivel(s)"

  def get_titulo(self):
    return self.__titulo

  def fazer_emprestimo(self):
    if self.__quantidade > 0:
      self.__quantidade -= 1
      return True
    return False

  def fazer_devolucao(self):
    self.__quantidade += 1


class Usuario:
  def __init__(self, nome, matricula):
    self.__nome = nome
    self.__matricula = matricula
    self.__livros_emprestados = []

  def __str__(self):
    return f"Usuario: {self.__nome}, Matricula: {self.__matricula}"

  def emprestar(self, livro):
    if livro.fazer_emprestimo():
      self.__livros_emprestados.append(livro)
      print(f"Livro {livro.get_titulo()} emprestado")
    else:
      print(f"Livro {livro.get_titulo()} indisponivel")

  def devolver(self, livro):
    if livro in self.__livros_emprestados:
      livro.fazer_devolucao()
      self.__livros_emprestados.remove(livro)
      print(f"O livro {livro.get_titulo()} foi devolvido")
    else:
      print(f"O livro {livro.get_titulo()} não consta como emprestado")

  def mostrar_livros_emprestados(self):
    titulos = [livro.get_titulo() for livro in self.__livros_emprestados]
    print(f"Livros de {self.__nome}: {titulos}")
````

````python
livro1 = Livro("O Pequeno Príncipe", "Saint-Exupéry", 3)
livro2 = Livro("1984", "George Orwell", 2)
````

````python
usuario1 = Usuario("Alice", "2023A")
usuario2 = Usuario("Bruno", "2023B")
````

````python
print(livro1)
print(livro2)
````

````python
usuario1.emprestar(livro1)
usuario1.emprestar(livro2)
usuario2.emprestar(livro1)
usuario2.emprestar(livro1)
usuario2.emprestar(livro1)
usuario2.emprestar(livro1)
````

````python
usuario1.mostrar_livros_emprestados()
usuario2.mostrar_livros_emprestados()
usuario1.devolver(livro1)
usuario1.mostrar_livros_emprestados()
print(livro1)
````

**Projeto 2: Gerenciamento de Conta Bancária com Herança**
<br>
**Contexto do mundo real:** Um banco que oferece dois tipos de conta: conta corrente e conta poupança.Queremos um sistema que permita depositar, sacar e aplicar juros (no caso da poupança).
<br>
**Descrição do problema:** Precisamos modelar contas bancárias em Python de forma orientada a objetos.
Teremos:
* Classe base Conta com atributos comuns (titular, saldo) e métodos depositar(valor) e sacar(valor).
* Classe ContaCorrente que herda de Conta e adiciona, por exemplo, taxa de saque (ou limite especial).
* Classe ContaPoupanca que herda de Conta e tem um método extra render_juros() que aumenta o saldo de acordo com uma taxa fixa anual.
Também podemos manter atributos de classe (como juros = 0.05 para 5% ao ano).

````python
class Conta:
  def __init__(self, titular, saldo=0):
    self.__titular = titular
    self.__saldo = saldo

  def depositar(self,valor):
    self.__saldo += valor

  def saque(self, valor):
    if self.__saldo <= valor:
      self.__saldo -= valor
    else:
      return "Não é possivel realizar o saque"

  def get_saldo(self):
    return self.__saldo

  def __str__(self):
    return f"{self.__titular}: Saldo R${self.__saldo:.2f}"
````

````python
class ContaCorrente(Conta):
  def __init__(self, titular, saldo_inicial=0, limite=0):
    super().__init__(titular, saldo_inicial)
    self.__limite = limite

  def sacar(self, valor):
    if 0 < valor <= self.get_saldo() + self.__limite:
      novo_saldo = self.get_saldo() - valor
      super()._Conta__saldo = novo_saldo
      if novo_saldo < 0:
        print(f"Saque de {valor:.2f} realizado (incluindo uso do limite).")
      else:
        print(f"Saque de {valor:.2f} realizado.")
    else:
      print("Saque não autorizado: limite excedido.")
````

````python
class ContaPoupanca(Conta):

  def __init__(self, titular, saldo_inicial=0, juros = 0.05):
    super().__init__(titular, saldo_inicial)
    self.__juros = juros

  def render_juros(self):
    rendimento = super().get_saldo() + (super().get_saldo() * self.__juros)
    super().depositar(rendimento)
    print(f"Juros de R${super().get_saldo() * self.__juros:0.2f} aplicados.")
````

````python
cc = ContaCorrente("Cliente CC", 1000.0, limite=500.0)
cp = ContaPoupanca("Cliente CP", 2000.0)
````

````python
print(cc)
print(cp)
cc.saque(1200)
print(cc)
cc.depositar(300)
print(cc)
cp.saque(500)
print(cp)
cp.render_juros()
print(cp)
````
