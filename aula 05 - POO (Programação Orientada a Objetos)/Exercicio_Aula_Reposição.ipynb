**Exercício 5 — Conta simples com encapsulamento [Fácil]**<br>
Crie uma classe Conta que mantenha o saldo protegido contra alteração direta.<br>
Requisitos:
* O construtor recebe o nome do titular e inicia o saldo em 0.
* Use __saldo como atributo privado.
* Implemente depositar(valor), permitindo apenas valores positivos.
* Implemente consultar_saldo(), retornando o saldo atual.
* Crie uma conta, faça dois depósitos e mostre o saldo.

```python
class Conta:
  def __init__(self, nome, saldo_inicial = 0):
    self.__titular = nome
    self.__saldo = saldo_inicial

  def depositar(self,valor):
    if valor > 0:
      self.__saldo += valor
    else:
      return "Valor invalido - Deposite um valor positivo"

  def consultar_saldo(self):
    return self.__saldo
```

```python
conta1 = Conta("Lucas")
conta1.depositar(200)
conta1.depositar(-16)
conta1.depositar(543)
conta1.consultar_saldo()
```

**Exercício 7 — Veículo com velocidade controlada [Média]**<br>
Modele um veículo inspirado no exemplo Automovel da aula, mas com regras completas de aceleração e frenagem.<br>
**Requisitos:**<br>
* A classe Veiculo deve ter placa, velocidade_max e velocidade_atual; torne os três atributos privados.
* velocidade_atual começa em 0.
* Crie getters para placa e velocidade_max.
* Crie um setter para velocidade_max que aceite apenas valores positivos e nunca deixe a nova máxima
menor que a velocidade atual.
* acelerar(incremento=10) aumenta a velocidade sem ultrapassar a máxima.
* frear(decremento=10) reduz a velocidade sem ficar negativa.
* Implemente __str__ para mostrar placa, velocidade atual e máxima.
* Demonstre o funcionamento com várias acelerações e frenagens.

```python
class Veiculo:
  def __init__(self,placa,velocidade_max,velocidad_atual=0):
    self.__placa = placa
    self.__velocidade_max = velocidade_max
    self.__velocidade_atual = velocidad_atual

  def getter_placa(self):
    return self.__placa

  def getter_velocidade_max(self):
    return self.__velocidade_max

  def setter_velocidade_max(self,valor):
    if valor > 0 and (valor + self.__velocidade_max) > self.__velocidade_atual:
      self.__velocidade_max += valor

  def acelerar(self, valor = 10):
    if self.__velocidade_max >= (self.__velocidade_atual + valor):
      self.__velocidade_atual += valor
    else:
      return "Velocidade excedeu o limite maximo"

  def freiar(self, valor):
    if (self.__velocidade_atual - valor) < 0:
      return "Velocidade Atual não pode ficar negativa"

  def __str__(self):
    return f"Placa: {self.__placa}\nVelocidade Atual: {self.__velocidade_atual}\nVelocidade Maxima: {self.__velocidade_max}\n"
```

```python
carro = Veiculo("ABC-1D34", 230)
carro.acelerar()
print(carro)
carro.acelerar(50)
carro.freiar(20)
print(carro)
```

**Exercício 8 — Funcionários e cálculo de bônus [Média]**<br>
Crie uma hierarquia para calcular bônus de funcionários de uma empresa.<br>
**Requisitos:**
* Funcionario possui nome e salario; implemente __str__.
* Funcionario.calcular_bonus() retorna 5% do salário.
* Vendedor herda de Funcionario, acrescenta vendas e redefine calcular_bonus() para 5% do salário + 2% das vendas.
* Gerente herda de Funcionario, acrescenta equipe e redefine calcular_bonus() para 10% do salário + R$100 por integrante da equipe.
* Crie pelo menos um objeto de cada tipo, coloque-os em uma única lista e exiba nome, tipo e bônus usando o mesmo método calcular_bonus().

```python
class Funcionario:
  def __init__(self, nome, salario):
    self._nome = nome
    self._salario = salario

  def __str__(self):
    return f"{self.__nome}: R${self.__salario}"

  def Calcular_bonus(self):
    return f"Nome: {self._nome}, cargo: Funcionario, Bonus: {self._salario * 0.05}"

class Vendendor(Funcionario):
  def __init__(self, nome, salario,venda=0):
    super().__init__(nome, salario)
    self.__venda = venda

  def Realizar_venda(self, venda):
    self.__venda += venda

  def Calcular_bonus(self):
    return f"Nome: {self._nome}, Cargo: Vendedor, Bonus: {self._salario * 0.05 + self.__venda * 0.02}"

class Gerente(Funcionario):
  def __init__(self, nome, salario,equipe=0):
    super().__init__(nome, salario)
    self.__equipe = equipe

  def Adicionar_integrante(self,valor=1):
    self.__equipe += valor

  def Calcular_bonus(self):
    return f"Nome: {self._nome}, cargo: Gerente, Bonus: {self._salario * 0.1 + (100 * self.__equipe)}"
```

```python
f1 = Funcionario("Lucas", 1800)
f2 = Vendendor("Rafael", 2500)
f3 = Gerente("Roberto", 4500)
```

```python
f2.Realizar_venda(30000)
f2.Realizar_venda(25000)
f2.Realizar_venda(5000)
f3.Adicionar_integrante()
f3.Adicionar_integrante()
```

```python
lista_funcionario = [f1,f2,f3]
```

```python
for f in lista_funcionario:
  print(f.Calcular_bonus())
```

Crie uma classe Produto para um catálogo de e-commerce, explorando atributo de classe e encapsulamento.<br>
**Requisitos:**
* Crie o atributo de classe contador iniciado em 0.
* Cada produto deve receber automaticamente um id sequencial a partir do contador.
* Use atributos privados para id, nome e preco.
* Crie getters para id, nome e preco.
* Crie set_preco(novo_preco), aceitando apenas valores maiores que zero.
* Implemente aplicar_desconto(percentual), validando percentual entre 0 e 100.
* Implemente __str__.
* Crie três produtos e, ao final, mostre quantos objetos foram criados usando Produto.contador.

```python
class Produto:
  contador = 0

  def __init__(self, nome, preco, id = contador):
    Produto.contador += 1
    self.__id = Produto.contador
    self.__nome = nome
    self.__preco = preco


  def get_id(self):
    return self.__id

  def get_nome(self):
    return self.__nome

  def get_preco(self):
    return self.__preco

  def set_preco(self, novo_preco):
    if novo_preco > 0:
      self.__preco = novo_preco

  def aplicar_desconto(self, desconto):
    if desconto > 0 and desconto < 100:
      percentual = desconto / 100
      self.__preco -= percentual * self.__preco

  def __str__(self):
    return f"ID: {self.__id}, Nome: {self.__nome}, Preço: {self.__preco}"

  def Calcular_frete(self):
    return 0
```

**Exercício 10 — Mini sistema de pedidos de e-commerce [Difícil]**<br>
Desenvolva um pequeno sistema orientado a objetos para pedidos de um e-commerce. O objetivo é integrar a maior parte dos conceitos trabalhados na aula em um único problema.<br>
**Requisitos:**
* Crie uma classe Produto com id automático (atributo de classe), nome e preço privados; inclua getters,setter validado para preço e __str__.
* Crie ProdutoFisico herdando de Produto, com peso, e ProdutoDigital herdando de Produto, com tamanho_mb.
* Em Produto, crie calcular_frete() retornando 0. Redefina-o em ProdutoFisico para retornar R$ 5,00 + R$2,00 por kg. ProdutoDigital deve manter frete igual a 0. Essa chamada comum deve demonstrar polimorfismo.
* Crie uma classe Cliente com nome e email privados, getters e __str__.
* Crie uma classe Pedido com número automático (atributo de classe), cliente, lista de produtos e status privado.
* Pedido deve oferecer adicionar_produto(produto), calcular_total(), calcular_frete_total(), fechar_pedido() e __str__.
* Depois de fechado, o pedido não deve aceitar novos produtos. Essa regra deve ser controlada
internamente pela classe.
* calcular_total() deve somar os preços dos produtos. calcular_frete_total() deve chamar calcular_frete() de
cada item sem testar se ele é físico ou digital.
* No programa principal, crie um cliente, ao menos três produtos (misturando físicos e digitais), monte um pedido, exiba o resumo, feche-o e tente adicionar mais um produto para comprovar a regra.


```python
class Produto:
  contador = 0

  def __init__(self, nome, preco, id = contador):
    Produto.contador += 1
    self.__id = Produto.contador
    self.__nome = nome
    self.__preco = preco


  def get_id(self):
    return self.__id

  def get_nome(self):
    return self.__nome

  def get_preco(self):
    return self.__preco

  def set_preco(self, novo_preco):
    if novo_preco > 0:
      self.__preco = novo_preco

  def aplicar_desconto(self, desconto):
    if desconto > 0 and desconto < 100:
      percentual = desconto / 100
      self.__preco -= percentual * self.__preco

  def __str__(self):
    return f"ID: {self.__id}, Nome: {self.__nome}, Preço: {self.__preco}"

  def Calcular_frete(self):
    return 0
```

```python
class ProdutoFisico(Produto):
  def __init__(self, nome, preco,peso, id=Produto.contador):
    super().__init__(nome, preco, id=Produto.contador)
    self.__peso = peso

  def Calcular_frete(self):
    return 5 + (2 * self.__peso)

class ProdutoDigital(Produto):
  def __init__(self, nome, preco, tamanho_mb, id=Produto.contador):
    super().__init__(nome, preco, id=Produto.contador)
    self.__tamanho_mb = tamanho_mb

  def Calcular_frete(self):
    return super().Calcular_frete()
```

```python
class Cliente:
  def __init__(self, nome, email):
    self.__nome = nome
    self.__email = email

  def get_nome(self):
    return self.__nome

  def get_email(self):
    return self.__email

  def set_nome(self, nome):
    self.__nome = nome

  def set_email(self, email):
    self.__email = email

  def __str__(self):
    return f"Nome: {self.__nome} - Email: {self.__email}"
```

```python
class Pedido:
  contador = 0

  def __init__(self, cliente, status = "aberto", lista_produto = [], id = contador):
    Pedido.contador += 1
    self.__id = Pedido.contador
    self.__cliente = cliente
    self.__status = status
    self.__lista_produto = []
    self.__valor_compra = 0
    self.__fretetotal = 0
    self.__valor_total = 0

  def adicionar_produto(self, produto):
    if self.__status == "aberto":
      print(produto)
      self.__lista_produto.append(produto)
    else:
      return "Pedido Fechado"

  def calcular_total(self):
    for p in self.__lista_produto:
      self.__valor_compra += p.get_preco()


  def calcular_frete_total(self):
    for p in self.__lista_produto:
      self.__fretetotal += p.Calcular_frete()

  def mostrar_produtos(self):
    for p in self.__lista_produto:
      print(p.get_nome())

  def fechar_pedido(self):
    self.__status = "fechado"
    self.calcular_frete_total()
    self.calcular_total()
    self.__valor_total = self.__fretetotal + self.__valor_compra
    print(f"""Cliente -- {self.__cliente.get_nome()} --
-- Produtos --""")
    self.mostrar_produtos()
    print(f"""-- Frete: {self.__fretetotal:,.2f} --
    -- Valor Produto: {self.__valor_compra:,.2f} --
    -- Valor Total: {self.__valor_total:,.2f} --\n""")

  def __str__(self):
    return f"Cliente: {self.__cliente}\n{self.__lista_produto}"
```

```python
p1 = ProdutoFisico("Tv Philips",5000, 4)
p2 = ProdutoFisico("Xiaomi 18 Pro Max", 7000, 0.343)
p3 = ProdutoFisico("Notebook Gamer", 10000, 4)
p4 = ProdutoDigital("Far Cry 7",300,5000000)
p5 = ProdutoDigital("Livro Estrutura de dados com Python.", 30,5)
```

```python
c1 = Cliente("Tuco","tuco@tuco.com.br")
c2 = Cliente("Rica","rica@outlook.com")
```

```python
pedido1 = Pedido(c1)
pedido2 = Pedido(c2)
```

```python
pedido1.adicionar_produto(p1)
pedido1.adicionar_produto(p3)
pedido1.adicionar_produto(p5)
pedido2.adicionar_produto(p2)
pedido2.adicionar_produto(p4)
```
