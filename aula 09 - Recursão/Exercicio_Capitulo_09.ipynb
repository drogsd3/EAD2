3. **(PRÁTICA)** Escreva uma função recursiva em Python
chamada soma_digitos(n) que calcule a soma dos dígitos de
um número inteiro positivo n. Por exemplo, soma_digitos(123)
deve retornar 1 + 2 + 3 = 6.

```python
def soma_digitos(n, soma=0):
    if n == 0:
        return soma
    else:
        return soma_digitos(n // 10, soma + n % 10)
```

```python
soma_digitos(123)
```
