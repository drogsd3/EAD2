4 - **(Teórica)** Sobre altura e nível em uma árvore, assinale a opção correta:<br>
A. A raiz sempre está no nível 1.<br>
B. A altura de uma árvore vazia é 1.<br>
C. O nível de um nó é a quantidade de filhos que ele possui.<br>
D. A altura de uma árvore é o maior nível em que há um nó.<br>
E. Em uma árvore, todos os nós têm a mesma altura.<br>
R: D

````python
#Codigo criados em aula
class No:
 def __init__(self, valor):
  self.valor = valor
  self.esquerda = None
  self.direita = None
````

````python
# Árvore do exemplo do capítulo
raiz = No('a')
raiz.esquerda = No('b')
raiz.direita = No('c')
raiz.esquerda.esquerda = No('d')
raiz.esquerda.direita = No('e')
````

5 - **(Prática)** Escreva uma função altura(raiz) em Python que recebe a raiz de uma árvore binária e retorna a altura da árvore (número de níveis). Dica: use recursão.<br>
Em seguida, teste-a na árvore criada no exemplo do capítulo.


````python
def altura(no):
  if no is None:
    return 0

  altura_esquerda = altura(no.esquerda)
  altura_direita = altura(no.direita)
  return max(altura_esquerda, altura_direita) + 1
````

````python
#Teste Saida esperada 3
print(altura(raiz))
````

6 - **(Prática)** Implemente em Python uma função conta_folhas(raiz) que receba a raiz de uma árvore binária e retorne o número total de nós folha.<br>
Mostre a chamada da função no exemplo do capítulo.


````python
def conta_folhas(no):
  if no is None:
    return 0

  if no.esquerda is None and no.direita is None:
    return 1

  return conta_folhas(no.esquerda) + conta_folhas(no.direita)
````

````python
#Teste Saida espera 3
print(conta_folhas(raiz))
````
