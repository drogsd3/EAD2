Retorne o índice do alvo no vetor caso seja encontrado, ou -1 caso contrário.

**Projeto 1: Busca Eficiente em um Catálogo de Produtos Online**
<br><br>
**Contexto:** Imagine que você está desenvolvendo a funcionalidade de busca para um site de e-commerce com um grande número de produtos. O catálogo de produtos é armazenado em uma lista, onde cada item é um dicionário contendo informações como nome, preço e descrição. Para otimizar a busca, o catálogo é mantido ordenado alfabeticamente pelo nome do produto.
<br><br>
**Problema:** Os clientes precisam encontrar rapidamente informações sobre um produto específico digitando seu nome na barra de busca. Uma busca lenta pode levar à frustração e à perda de vendas.
<br><br>
**Indicação de como resolver:** Implemente um sistema de busca que utilize a busca binária para localizar o produto no catálogo ordenado. Ao receber o nome do produto digitado pelo usuário, sua função deve realizar a busca binária na lista de produtos. Se o produto for encontrado, retorne suas informações (por exemplo, preço, descrição). Caso contrário, informe que o produto não foi encontrado.

```python
lista_produto = [
 {"nome": "Câmera", "preco": 1899.90, "descricao": "Câmera digital"},
 {"nome": "Fone", "preco": 249.90, "descricao": "Fone sem fio"},
 {"nome": "Notebook", "preco": 4599.00, "descricao": "Notebook 15 polegadas"},
 {"nome": "Smartphone", "preco": 2299.00, "descricao": "Smartphone 5G"},
 {"nome": "Tablet", "preco": 1599.00, "descricao": "Tablet 10 polegadas"}
]
```

```python
def buscar_produto(lista_produto, busca):
  dir = len(lista_produto) - 1
  esq = 0
  while dir >= esq:
    meio = (dir + esq) //2
    if busca == lista_produto[meio]["nome"]:
      return lista_produto[meio]
    elif busca < lista_produto[meio]["nome"]:
      dir = meio - 1
    else:
      esq = meio + 1
  return -1
```

```python
busca = input("Digite o nome do produto que deseja buscar: ")
produto = buscar_produto(lista_produto, busca)
if produto == -1:
  print("Produto não encontrado")
else:
  print(f"Produto encontrado: {produto['nome']}")
  print(f"Preço: {produto['preco']}")
  print(f"Descrição: {produto['descricao']}")
```

**Projeto 2: Busca Eficiente em um Índice de Livros**
<br><br>
**Contexto:** Você está trabalhando em um projeto para gerenciar uma biblioteca. As informações sobre os livros (título, autor, ano de publicação) estão armazenadas em uma lista, e essa lista é mantida ordenada alfabeticamente pelo título do livro.
<br><br>
**Problema:** Os usuários precisam encontrar rapidamente um livro específico sabendo seu título. Uma busca ineficiente pode dificultar a localização dos livros desejados.
<br><br>
**Indicação de como resolver:** Desenvolva uma funcionalidade que permita aos usuários buscar um livro pelo título utilizando a busca binária. Ao receber o título do livro, sua função deve procurar no índice ordenado. Se o livro for encontrado, exiba suas informações (título, autor, ano). Caso contrário, informe que o livro não foi
encontrado.

```python
livro_biblioteca = [
 {"titulo": "1984", "autor": "George Orwell", "ano": 1949},
 {"titulo": "Dom Casmurro", "autor": "Machado de Assis", "ano": 1899},
 {"titulo": "O Pequeno Príncipe", "autor": "Antoine de Saint-Exupéry", "ano": 1943},
 {"titulo": "Orgulho e Preconceito", "autor": "Jane Austen", "ano": 1813},
 {"titulo": "Vidas Secas", "autor": "Graciliano Ramos", "ano": 1938}
]
```

```python
def buscar_livro(livro_biblioteca, busca):
  dir = len(livro_biblioteca) - 1
  esq = 0
  while dir >= esq:
    meio = (dir + esq) //2
    if busca == livro_biblioteca[meio]["titulo"]:
      return livro_biblioteca[meio]
    elif busca < livro_biblioteca[meio]["titulo"]:
      dir = meio - 1
    else:
      esq = meio + 1
  return -1
```

```python
busca = input('Digite o título do livro que deseja buscar: ')
livro = buscar_livro(livro_biblioteca, busca)
if livro == -1:
  print("Livro não encontrado")
else:
  print(f"Livro encontrado: {livro['titulo']}")
  print(f"Autor: {livro['autor']}")
  print(f"Ano: {livro['ano']}")
```
