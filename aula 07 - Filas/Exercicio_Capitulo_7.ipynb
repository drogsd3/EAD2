Para a parte pratica já foi estabelecidade em aula a Class QUEUE (fila), seu atributos e metodos, que estão abaixo.

```python
# Implementação da classe Fila
class Queue:
    # Inicia com uma fila vazia
    def __init__(self):
        self.itens = []

    # Verifica se fila está vazia
    def is_empty(self):
        return self.itens == []

    # Adiciona elemento no início da fila
    def enqueue(self, item):
        self.itens.insert(0, item)

    # Remove elemento do final da fila
    def dequeue(self):
        return self.itens.pop()

    # Retorna o número de elementos da fila
    def size(self):
        return len(self.itens)

    # Imprime a fila na tela
    def print_queue(self):
      print(self.itens)
```

4. **(Exercicio Pratico)** - Fila de Impressão em um Escritório.<br>
Implemente um sistema de gerenciamento de fila de impressão. Cada documento tem um nome. Os documentos são processados na ordem em que chegam. Após a impressão de cada três documentos, a impressora precida de uma pausa para "resfriamento", durante a qual nenhum novo documento pode ser processado por um ciclo(mas novos documentos ainda podem ser adicionados à fila).

```python
def fila_de_impressao(documento):
    fila = Queue()
    for doc in documento:
        fila.enqueue(doc)
        documentos_impressos = 0
        ciclo = 0
    while not fila.is_empty():
        if documentos_impressos % 3 == 0 and ciclo == 0:
          print("Pausa para o resfriamento da impressora.")
          ciclo = 1
          continue
        doc_atual = fila.dequeue()
        print(f"Imprimindo documento: {doc_atual}")
        documentos_impressos += 1
        ciclo = 0
```

```python
fila_de_impressao(["A", "B", "C", "D", "E", "F", "G", "H", "I"])
```

5. (Exercicio Pratico) - Gerenciador de Senhas de Atendimento (Senha Preferencial e Comum). Implemente um sistema de atendimento com duas filas: uma para senhas preferenciais e outra para senhas comuns. A cada ciclo, atenda primeiro uma senha preferencial (se houver), depois uma senha comum. Repita até que ambas as filas estajam vazias.

```python
def gerenciar_atendimento(preferenciais, comuns):
    fila_preferencial = Queue()
    fila_comum = Queue()

    for senha in preferenciais:
        fila_preferencial.enqueue(senha)
    for senha in comuns:
        fila_comum.enqueue(senha)
    while not fila_preferencial.is_empty() or not fila_comum.is_empty():
        if not fila_preferencial.is_empty():
            senha_atendida = fila_preferencial.dequeue()
            print(f"Atendendo Preferencial: {senha_atendida}")
        if not fila_comum.is_empty():
            senha_atendida = fila_comum.dequeue()
            print(f"Atendendo Comum: {senha_atendida}")
```

```python
gerenciar_atendimento(["P1", "P2", "P3"], ["C1", "C2", "C3", "C4"])
```
